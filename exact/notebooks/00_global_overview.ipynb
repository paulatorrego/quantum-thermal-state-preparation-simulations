# TFM — Preparación de estados térmicos cuánticos mediante baños auxiliares, medidas y resets

Estudio de protocolo de preparación térmica cuántica basados en interacciones repetidas entre un sistema y baños auxiliares en distintos modelos. La idea común es reemplazar una evolución puramente unitaria por un canal disipativo implementable: se inicializan ancillas, se acoplan al sistema durante una ventana temporal, se miden o resetean, y se repite el proceso hasta aproximar un estado de Gibbs.

El trabajo no busca demostrar una ventaja experimental inmediata, sino construir una cadena progresiva de validación física, numérica y circuital del protocolo. Además de comprobar la convergencia hacia estados de Gibbs en modelos pequeños, se estudian los mecanismos que explican cuándo el protocolo funciona y cuándo falla:

1. validación en modelos exactamente resolubles;
2. análisis de convergencia hacia Gibbs y del sesgo del punto fijo;
3. estudio de operadores de acoplo y de las simetrías que rompen o conservan;
4. análisis espectral de gaps de Bohr y transiciones energéticas accesibles;
5. identificación de resonancias asociadas a la ventana temporal del filtro;
6. comparación de protocolos con y sin randomización;
7. análisis del rewind, incluyendo casos donde no mejora el resultado;
8. estudio de tiempos de mezcla y búsqueda negativa de efecto Mpemba;
9. extensión progresiva a modelos interactuantes como Ising y Fermi-Hubbard;
10. análisis de sectores físicos, especialmente half-filling en Fermi-Hubbard;
11. traducción a circuitos Qiskit dinámicos con ancillas, medidas y resets;
12. construcción de estimadores shot-based para observables diagonales y no diagonales;
13. reconstrucción de energía total mediante Pauli strings;
14. estimación de recursos, profundidad, puertas de dos qubits y coste ISA;
15. comparación ideal/noisy y sensibilidad a ruido calibrado.

# Papers de referencia 

## Quantum thermal state preparation for near-term quantum processors  (Abanin & Lloyd)

Este es el paper central. Propone un protocolo de preparación térmica para procesadores near-term basado en baños auxiliares reseteados y acoplos sistema-baño modulados temporalmente. La motivación principal es obtener un canal cuya distribución estacionaria aproxime el estado de Gibbs del sistema,

$$
\rho_\beta \propto e^{-\beta H_S}.
$$

El paper predice que, en régimen de acoplo débil, el error del punto fijo escala aproximadamente como

$$
\|\rho_\infty-\rho_\beta\| \sim \theta^2,
$$

donde $\theta$ controla la intensidad del acoplo sistema-baño. Además, estudia modelos de Ising 2D y sistemas free-fermion como ejemplos numéricos relevantes.

Se usa como base de la construcción principal. A partir de él se implementan los protocolos de baños repetidos en NB01, NB02, NB03 y NB04. La estructura de ciclos, la modulación temporal, el papel de $M_T$, $\delta$, $\theta$, los resets de baños y el análisis de punto fijo vienen de esta referencia.

Resultados obtenidos en relación con este paper:

- en NB01 se reproduce el comportamiento esperado en el caso mínimo de un qubit;
- en NB02 se comprueba que el protocolo escala de forma controlada en un modelo no interactuante;
- en NB03 se extiende el estudio a Ising 1D y 2D, incluyendo correlaciones y geometría;
- en NB03 también se incluye un apéndice free-fermion como conexión con el régimen integrable tratado en el paper;
- en NB04 se adapta la filosofía del protocolo a Fermi-Hubbard mediante Jordan-Wigner;
- se observa que el régimen débil/intermedio es el más estable, mientras que los regímenes resonantes o de acoplo fuerte requieren retuning.

## Provably Efficient Quantum Thermal State Preparation via Local Driving (Hahn) 

Este paper aporta la parte más rigurosa y algorítmica. Estudia preparación térmica mediante driving local, acoplos locales dependientes del tiempo y resets de ancillas, con garantías de eficiencia bajo hipótesis de localidad y mezcla.

Su importancia para este TFM es que justifica que no hace falta construir baños macroscópicos ni operadores completamente no locales para aproximar estados térmicos. La preparación puede formularse mediante ingredientes más cercanos a hardware real:

- simulación del Hamiltoniano del sistema;
- acoplos locales sistema-ancilla;
- driving temporal;
- resets repetidos.

En este TFM se usa como marco conceptual para mantener los acoplos locales y estudiar implementaciones que no dependan de conocer todo el espectro de forma perfecta. También motiva la comparación entre protocolos tipo `FullBath` y `Sequential`, especialmente en Fermi-Hubbard.

Resultados obtenidos en relación con este paper:

- en NB02 y NB03 se prueban acoplos locales y geometrías de baño distintas;
- en NB04 se comparan variantes `FullBath` y `Sequential`, donde `Sequential` reutiliza baños y reduce coste conceptual;
- los mapas en $(\theta, M_T)$ muestran que existen ventanas locales de buena preparación, pero también resonancias;

## Thermal state preparation by repeated interactions at and beyond the Lindblad limit (Carlos Ramon-Escandell)

Este paper estudia protocolos de interacciones repetidas más allá del límite Lindblad. Analiza cómo cambia la termalización cuando las colisiones sistema-ancilla no se toman en el límite infinitesimal y cómo depende el tiempo de preparación del número de interacciones. También discute efectos tipo Mpemba, donde estados inicialmente más calientes pueden tardar más o menos en alcanzar el equilibrio dependiendo del régimen.

En este TFM se usa para motivar el análisis de tiempos de mezcla y la interpretación de los ciclos como colisiones repetidas. También sirve como referencia para no limitar el estudio al régimen Lindblad ideal, sino explorar directamente mapas discretos de tiempo finito.

Resultados obtenidos en relación con este paper:

- en NB01 se analiza cómo el número de ciclos controla la aproximación a Gibbs;
- en NB03 se estudian tiempos de mezcla para Ising y se busca una señal de efecto Mpemba;
- en los casos estudiados de Ising no aparece una señal robusta y universal de Mpemba;
- sí se observa una fuerte dependencia no monótona con parámetros como $\theta$, $M_T$, $\delta$ y la estructura espectral;
- esto confirma que, fuera del límite Lindblad, la elección de ventana temporal y acoplo es tan importante como la temperatura objetivo.

---

## Stable Quantum-Correlated Many Body States through Engineered Dissipation (Google)

Este paper es la referencia experimental y de hardware. Demuestra en procesadores superconductores que la disipación ingenierizada con qubits auxiliares puede preparar estados correlacionados de muchos cuerpos, especialmente en modelos tipo Ising. Su mensaje principal es que la disipación no es solo una fuente de error, sino también una herramienta para preparar estados cuánticos de interés.

En este TFM se usa como motivación para llevar los protocolos a circuitos Qiskit dinámicos con medidas, resets y ancillas. También justifica estudiar no solo la convergencia exacta, sino el coste de implementación, los recursos de circuito y la degradación por ruido.

Resultados obtenidos en relación con este paper:

- en NB01, NB02, NB03 y NB04 se construyen circuitos dinámicos con ancillas y resets;
- en NB03 se estudia explícitamente Ising 1D/2D y la aparición de correlaciones;
- en NB03 y NB04 se estiman recursos de circuitos, profundidad, puertas de dos qubits y coste ISA;
- se comparan ejecuciones ideales y noisy cuando es viable;
- el resultado práctico es que la física del protocolo es coherente, pero el coste y el ruido acumulado limitan rápidamente el tamaño ejecutable en hardware NISQ.

# Estructura de notebooks

## NB01 — Single spin

Este notebook estudia el caso mínimo: un sistema de un qubit acoplado repetidamente a baños auxiliares. Sirve para validar la física básica del protocolo sin complicaciones de tamaño o interacción.

Se analiza:

- convergencia hacia Gibbs;
- dependencia con $\theta$, $M_T$, $\beta$, $h$ y $\delta$;
- resonancias del filtro temporal;
- efecto de randomización y rewind;
- implementación Qiskit ideal;
- estimadores shot-based;
- ruido calibrado y ruido artificial.

Resultado principal: el protocolo reproduce la termalización esperada en el caso mínimo y permite identificar de forma clara el régimen perturbativo, las resonancias y el papel del número de ciclos.

---

## NB02 — Modelo no interactuante

Este notebook extiende el single spin a varios qubits independientes. Es un puente entre el caso mínimo y los modelos interactuantes.

Se analiza:

- factorización del Gibbs;
- escalado con el número de qubits;
- geometrías de baño;
- resonancias heredadas del single spin;
- ausencia de correlaciones físicas genuinas;
- circuitos Qiskit dinámicos;
- estimadores shot-based y ruido.

Resultado principal: el modelo no interactuante permite separar efectos de tamaño de efectos de interacción. La preparación térmica sigue siendo interpretable y controlada, pero el coste de circuito ya empieza a crecer.

---

## NB03 — Ising 1D y 2D

Este notebook introduce interacciones $XX$ y geometría. Es el primer caso donde aparecen correlaciones físicas no triviales.

Se estudian:

- Ising 1D y 2D exactos;
- comparación entre geometrías;
- operadores de acoplo;
- geometría del baño;
- resonancias y randomización;
- tiempos de mezcla;
- análisis negativo de efecto Mpemba;
- apéndice free-fermion como régimen integrable;
- validación Qiskit ideal;
- reconstrucción de energía total combinando bases $Z$ y $X$;
- ruido ideal vs noisy.

Resultado principal: el protocolo sigue funcionando en presencia de interacciones, pero se vuelve más sensible a la estructura espectral, al ruido y al coste de implementación. La energía total requiere medir observables no diagonales, lo que aumenta el coste shot-based.

---

## NB04 — Fermi-Hubbard

Este notebook estudia el caso más rico del proyecto: el modelo Fermi-Hubbard, empezando por el dímero y extendiendo a tamaños pequeños.

Se analiza:

- construcción Jordan-Wigner;
- simetrías y sectores de partículas;
- convergencia a Gibbs;
- operadores de acoplo;
- barridos en $U/t$, temperatura, $h$ y $M_T$;
- mapas en $(\theta, M_T)$;
- resonancias, randomización y rewind;
- escalado del protocolo;
- termodinámica exacta;
- variante canónica a half-filling;
- circuitos Qiskit `FullBath` y `Sequential`;
- energía diagonal por shots;
- reconstrucción de energía total por Pauli strings;
- ruido ideal vs noisy.

Resultado principal: Fermi-Hubbard confirma que el protocolo puede aproximar estados térmicos en regímenes débiles e intermedios, pero que en régimen Mott fuerte aparecen limitaciones claras. La variante canónica a half-filling ayuda a controlar el sector de partículas, y la comparación `FullBath`/`Sequential` muestra el compromiso entre fidelidad física y coste de implementación.

# Mensaje global del TFM

El resultado global es una validación progresiva de protocolos de preparación térmica por baños repetidos. Los modelos simples confirman la física esperada y permiten identificar resonancias y escalados perturbativos. Los modelos interactuantes muestran que la preparación sigue siendo posible, pero requiere ajustar cuidadosamente acoplos, ventanas temporales y geometrías de baño. La traducción a Qiskit demuestra que el protocolo puede formularse como circuitos dinámicos con ancillas y resets, aunque el coste y el ruido acumulado limitan el tamaño práctico en hardware NISQ.

En conjunto, el proyecto conecta cuatro niveles: teoría de preparación térmica, dinámica de interacciones repetidas, disipación ingenierizada en hardware y simulación numérica reproducible en modelos many-body pequeños.
